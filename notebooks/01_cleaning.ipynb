**`Landsize` min == 0**:

```python
pd.crosstab(df["Landsize"] == 0, df["Type"])
```

- Mostly exist in units `u` (1629), much greater than houses `h` (171) or townhouses `t` (139)

```python
pd.crosstab(df["Landsize"] == 0, df["Type"],  normalize="columns")
```

- Normalizing by column shows ~54% of units have a `Landsize` of 0 compared to ~1.8% for houses and ~12.5% for townhouses
    - This happens since units/aparentments often share land with a building so a 0 could be a "not applicable" filler since the entire floor could be owned and measured collectively
        - Hence I will be using the 0 values in my data analysis since it is arguably correct (no individual land rather than missing value)
    - For houses, the 2% rate means more likely it could ben an error or missing value
        - For houses and townhouses, by definition they cannot have 0m² of land.
        - Therefore for `Landsize` == 0 where `Type` are `h` or `t`, `Landsize` will be set to **`NaN`**

```python
df.loc[(df["Type"]=="u") & (df["Landsize"]>0), "Landsize"].describe()
```

- For units `u` with a landsize above 0, the distribution is right-skewed since median is ~1037, min is 1 and max is 37000
    - This gap means a small number of extreme outliers pulling the average upwards 
    - 37000m² is definitely not possible for a single unit, so it might be a recording of a whole complex/building rather than a single unit, which means there might be inconsistencies of what is classified and measured for `u`

**`Landsize` large max value**:

```python
landsize_top = df.loc[:, ["Suburb", "Type", "Price", "Landsize", "BuildingArea", "Regionname"]].sort_values("Landsize", ascending= False).head(20).reset_index(drop=False)
landsize_top.index+=1
landsize_top
```

- 2nd, 4th, 6th, 7th highest `Landsize` houses `h` have `Regionname` outside metropolitan areas
    - Probable since their locations are in sem-rural hill areas
- 3rd, 5th highest `Landsize` houses `h` have `Regionname` inside metropolitan areas
    - In metropolian areas, there are strict rules in place for extremely large sizes of private homes.
        - The largest recorded Balwyn North house (mega mansion) is ~1000 square meters
        - The largest recorded Reservoir house is ~500 square meters
    - If those were its real recorded `Landsize` the property's `Price` will be much higher than $2M or $500K
    - Hence for `Landsize` = 75100 and 41400 (index = 687, 5194), `Landsize` will be set to **`NaN`**

```python
print(f"Top 2 difference: +{landsize_top.loc[1, "Landsize"]-landsize_top.loc[2, "Landsize"]}, x{landsize_top.loc[1, "Landsize"]/landsize_top.loc[2, "Landsize"]}")
```

```python
print(f"99th percentile: {df["Landsize"].quantile(0.99)}")
print(f"Ratio to 1st item: x{df["Landsize"].max()/df["Landsize"].quantile(0.99)}")
```

- 1st (top) `Landsize` is 357014m² more than the 2nd highest `Landsize` 
    - This is 5 times the 2nd highest `Landsize`, which is 76000m²
    - It's area is ~146x the 99th percentile (2959.83)
    - Located in Fitzroy = ~1,000,000m² (accounting streets and boundaries), which a house `h` taking up 1/2 in one of the most densely populated suburbs is physically impossible at all
    - Hence for `Landsize` = 433014 (index = 11020), `Landsize` will be set to **`NaN`**

```python
## standard outlier rule: Q3 + 1.5*IQR, where IQR is Q3-Q1
landsize_Q1 = df["Landsize"].quantile(0.25)
landsize_Q3 = df["Landsize"].quantile(0.75)
landsize_IQR = landsize_Q3-landsize_Q1
landsize_upper_outlier = landsize_Q3+(1.5*landsize_IQR)

print(f"Upper Outlier: {landsize_upper_outlier}")
print(f"Amount of outliers w/Type 'u': {((df["Type"]=="u") & (df["Landsize"]>landsize_upper_outlier)).sum()}")
df.loc[(df["Type"]=="u") & (df["Landsize"]>landsize_upper_outlier),["Suburb", "Type", "Price", "Landsize", "BuildingArea", "Regionname"]].sort_values("Landsize", ascending= False)
```

```python
print(f"Rows with duplicated landsize: {df.loc[(df["Type"]=="u") & (df["Landsize"]>landsize_upper_outlier),"Landsize"].duplicated(keep = False).sum()}")
print(f"Percent of duplicated outliers: {(df.loc[(df["Type"]=="u") & (df["Landsize"]>landsize_upper_outlier),"Landsize"].duplicated(keep = False).sum())/(((df["Type"]=="u") & (df["Landsize"]>landsize_upper_outlier)).sum())*100}%")
```

- For units `u`, there were 248 properties with `Landsize` > 1362m² (upper outlier: Q3+1.5*IQR)
    - There are 70 duplicated units `u` that are above the upper outlier, or ~28% of the outliers
        - Duplicated values most likely mean they belong to the **same development** site
        - Meaning they are likely to represent the total land area of the development rather than individual units
        - Hence it is safer to keep the data as original rather than claiming an error
